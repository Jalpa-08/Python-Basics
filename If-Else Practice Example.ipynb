```python
# Common list elements
list1 = [10, 20, 30, 40, 50]
list2 = [30, 60, 10, 90]

common = list(set(list1) & set(list2))

if common:
    print(f"Common elements found: {common}")
    if sum(common) > 50:
        print("Sum of common elements is greater than 50.")
    else:
        print("Sum of common elements is 50 or less.")
else:
    print("No common elements found.")
```

```python
# Dictionary Student Grade Checker using for loop
grades = {
    "Alice": 85,
    "Bob": 67,
    "Charlie": 95,
    "David": 50
}

for student, grade in grades.items():
    if grade >= 90:
        print(f"{student} got A grade")
    elif grade >= 75:
        print(f"{student} got B grade")
    elif grade >= 60:
        print(f"{student} got C grade")
    else:
        print(f"{student} needs improvement")
```

```python
# Check if a key exists in a dictionary
data = {"name": "Alice", "age": 25}
key_to_check = "age"

if key_to_check == "name":
    print("Key exists:", data["name"])
else:
    if key_to_check == "age":
        print("Key exists:", data["age"])
    else:
        print("Key does not exist")
```

```python
#  Check if two dictionaries have the same values for a key
dict1 = {"id": 101, "score": 90}
dict2 = {"id": 102, "score": 90}

if "score" in dict1:
    if "score" in dict2:
        if dict1["score"] == dict2["score"]:
            print("Both have the same score.")
        else:
            print("Scores are different.")
    else:
        print("Key 'score' not in dict2")
else:
    print("Key 'score' not in dict1")
```

```python
# Program for a simple calculator
num1 = float(input("Enter first number: "))
op = input("Enter operator (+, -, *, /): ")
num2 = float(input("Enter second number: "))

if op == '+':
    print(f"Result: {num1 + num2}")
elif op == '-':
    print(f"Result: {num1 - num2}")
elif op == '*':
    print(f"Result: {num1 * num2}")
elif op == '/':
    if num2 != 0:
        print(f"Result: {num1 / num2}")
    else:
        print("Error: Division by zero is not allowed")
else:
    print("Invalid operator")
```

```python
# Program to check voting eligibility
age = int(input("Enter your age: "))
citizen = input("Are you a citizen? (yes/no): ").lower()

if age >= 18 and citizen == 'yes':
    print("You are eligible to vote")
elif age < 18:
    print("You are too young to vote")
elif citizen == 'no':
    print("You must be a citizen to vote")
else:
    print("Invalid input for citizenship")
```

```python
# Program to select a menu item from a dictionary
menu = {'burger': 10, 'pizza': 15, 'salad': 8, 'drink': 3}

choice = input("Enter menu item (burger/pizza/salad/drink): ").lower()

if choice in menu:
    quantity = int(input("Enter quantity: "))
    if quantity > 0:
        total = menu[choice] * quantity
        print(f"Total cost for {quantity} {choice}(s): ${total:.2f}")
    else:
        print("Invalid quantity")
else:
    print("Item not on the menu")
```

```python
# Program to check student grade using a dictionary
grades = {'Alice': 85, 'Bob': 92, 'Charlie': 78, 'Diana': 95}

name = input("Enter student name: ").capitalize()

if name in grades:
    score = grades[name]
    if score >= 90:
        print(f"{name}'s grade: A+")
    elif score >= 80:
        print(f"{name}'s grade: A")
    elif score >= 70:
        print(f"{name}'s grade: B")
    else:
        print(f"{name}'s grade: C or below")
else:
    print(f"No record found for {name}")
```
